#**3.4.1:**

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)

class1=np.random.randn(100,2)+[2,2]
class2=np.random.randn(100,2)+[-2,-2]
plt.scatter(class1[:,0],class1[:,1],label='Class 1')
plt.scatter(class2[:,0],class2[:,1],label='Class 2')
plt.legend()
plt.title("Linearly Seperable Data")
plt.show()
```

# **3.4.2:**

```python
theta = np.linspace(0, 2*np.pi, 100)
circle1 = np.c_[np.cos(theta), np.sin(theta)]
circle2 = 2 * circle1
plt.scatter(circle1[:,0], circle1[:,1], label='Class 1')
plt.scatter(circle2[:,0], circle2[:,1], label='Class 2')
plt.legend()
plt.title("Non-Linearly Separable Data")
plt.show()
```

#**3.4.3:**
Geometric difference: Concentric circles (radius 1 vs 2) where no straight line separates inner from outer ring.​


#**3.4.4:**

| Dataset Type           | Good Models                                             | Poor Models                     | Reason                                                        |
| ---------------------- | ------------------------------------------------------- | ------------------------------- | ------------------------------------------------------------- |
| Linearly Separable     | Logistic Regression, Linear SVM, Perceptron             | Most neural networks (overkill) | Simple hyperplane finds perfect separation                    |
| Non-Linearly Separable | RBF SVM, Decision Trees, Neural Networks, Random Forest | Linear SVM, Logistic Regression | Requires kernel trick or non-linear boundaries geeksforgeeks​ |



#Linear dataset:  
Achieves 100% accuracy with linear models.Non-linear dataset: Linear models fail (50% accuracy); kernel SVM reaches ~99% via implicit high-dimensional mapping
