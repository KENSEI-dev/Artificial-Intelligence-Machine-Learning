```python
from collections import deque
def reconstruct_path(parent, goal):
    path = []
    current = goal
    while current is not None:
        path.append(current)
        current = parent.get(current)
    return path[::-1]
def bfs_path(graph, start, goal):
    if start == goal:
        return [start], [start]

    queue = deque([start])
    visited = {start}
    parent = {start: None}
    sequence = [start]

    while queue:
        node = queue.popleft()
        sequence.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                parent[neighbor] = node
                if neighbor == goal:
                    path = reconstruct_path(parent, goal)
                    return path, sequence

                visited.add(neighbor)
                queue.append(neighbor)

    return None, sequence

museum = {
    'E': ['B', 'A'],
    'A': ['C', 'D'],
    'B': ['D'],
    'C': ['F'],
    'D': ['C', 'G'],
    'F': ['D'],
    'G': ['V'],
    'V': []
}


route, sequence =bfs_path(museum, 'C', 'V')


print(f"1. Room Exploration Sequence: {sequence}")
print(f"2. Shortest Path to Vault: {route}")
moves = len(route) - 1 if route else float('inf')
print(f"3. Total Moves Required: {moves}")

if moves <= 4:
    print("4. Outcome: SUCCESS — Vault secured within time limit!")
else:
    print("4. Outcome: FAILURE — Exceeded maximum allowed moves.")
```

8.1.4.
BFS is the most appropriate algorithm because,
i) Shortest Path Guarantee: The security system locks the vault after 4 moves. We want the goal with the minimum number of steps. BFS is mathematically proven to find the shortest path in unweighted graphs.
ii)Completeness: BFS will definitely find the solution if one exists.

8.1.5.
Yes, BFS guarantees the optimal solution.
BFS visits every node at depth d before visiting any other node.BFS will check all paths of length.  If V is not found at those depths, the first time BFS encounters V. After BFS it is impossible for any other shorter path exist.

8.1.6.
Complexity Analysis
Time: O(V + E), visits each node/edge once (V=8 rooms, E=10 corridors).
Space: O(V) for queue + visited + parent (worst-case: all nodes enqueued).
Scales linearly; efficient for museum-scale graphs.
